# Importing Libraries

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from keras.datasets import boston_housing
from keras.models import Sequential
from keras.layers import Dense
```

# Data Collection

```python
(x_train, y_train), (x_test, y_test) = boston_housing.load_data()
```

# Data Understanding

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

# Data Preparation

```python
standard_scale =StandardScaler()
```

```python
x_train = standard_scale.fit_transform(x_train)
```

```python
x_test = standard_scale.fit_transform(x_test)
```

# Model Building

```python
model = Sequential()
model.add(Dense(units = 64,activation = 'relu',input_shape = (13,)))
model.add(Dense(units=32,activation= 'relu'))
model.add(Dense(units=16,activation= 'relu'))
model.add(Dense(units= 1))
```

```python
model.compile(optimizer='adam',loss = 'mse',metrics = ['mae'])
```

# Model Training | Model Testing | Model Evaulate

```python
model.fit( x=x_train, y=y_train,batch_size=10,epochs=100,validation_split=0.20)
```

# Prediction

```python
data = np.array([[0.2,12,0.36,5,0.5,6,3.1,8,12,1.3,4.8,1.9,2]])
```

```python
scaling = standard_scale.transform(data)
scaling
```

```python
pred = model.predict(scaling)
pred
```

```python
print("Prediction ",pred)
```
