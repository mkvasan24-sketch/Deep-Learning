# importing neccessary libraries

```python
import keras
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
from keras.models import Sequential
from keras.layers import Flatten,Dense
```

# Data collection

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()
```

# Data understanding

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

# Data Preparation

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
x_train.min(),x_train.max()
```

# Model Building

```python
model = Sequential()   # building a sequential model
model.add(Flatten(input_shape = (32,32,3)))  # input layer must be 32*32*3 = 3072 neurons
model.add(Dense( units = 312,activation='relu')) # Hidden layer 1 
model.add(Dense(units= 256,activation='relu'))  # Hidden Layer 2
model.add(Dense(units= 128,activation= 'relu'))  # Hidden Layer 3 
model.add(Dense(units=10,activation = 'softmax')) # Output Layer
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

# Model Training | Model Evaulate | Model Testing

```python
model.fit(  x= x_train,y=y_train,batch_size=70,epochs=10,validation_split=0.20,verbose=2)
```

# Prediction 

```python
class_name =['airplane','automobile','bird','cat','deer','dog','frog','horse','ship','truck']
```

```python
y_pred = model.predict(x_test)
```
